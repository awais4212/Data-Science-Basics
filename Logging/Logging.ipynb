#### Python Logging

>Python’s logging module records messages about what a program is doing. It is better than using print() because you can control:

* **Which messages are shown**
* **Where messages are stored**
* **How messages are formatted**
* **The severity level of each message**

* **DEBUG (10):** Detailed diagnostic information for debugging.
* **INFO (20):** Normal program events confirming successful operation.
* **WARNING (30):** Unexpected situation that does not stop execution.
* **ERROR (40):** A failure that prevents a specific operation from completing.
* **CRITICAL (50):** Severe failure that may stop the entire application.


```python
import logging

# Basic Configuration
logging.basicConfig(level=logging.DEBUG)

# Log Messages With Severity level
logging.debug('This is a debug Message')
logging.info('This is an Info Message')
logging.warning('This is a Warning Message')
logging.error('This is an Error Message')
logging.critical('This is a Critical Message')
```

```python
# Basic Configuration and formatting
import logging
logging.basicConfig(
    filemode='w',
    filename='app.log',
    level = logging.DEBUG,
    format= '%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
logging.debug('This is a debug Message')
logging.info('This is an Info Message')
logging.warning('This is a Warning Message')
logging.error('This is an Error Message')
logging.critical('This is a Critical Message')
```

#### Logging With Multiple Logger 

```python
import logging

logger1 = logging.getLogger('module1')
logger1.setLevel(logging.DEBUG)

logger2 = logging.getlogger('module2') 
logger2.setlevel(logging.WARNING)

logging.basicConfig(
    level=logging.DEBUG,
    format = '%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt = '%Y-%m-%d %Y:%M:%S'
)
```
