```python
## Counting the Words, Lines & Characters
def count_alphabet_words(file_path):
    with open (file_path, mode='r') as file:
        lines = file.readlines()
        linesCount = len(lines)
        wordCount = sum(len(line.split()) for line in lines)
        charCount = sum(len(line) for line in lines)
    return linesCount, wordCount, charCount


lines, words, character = count_alphabet_words('Awais.txt')
print(f"The Lines are: {lines}, words are {words}, characters are {character}")
```

```python
from abc import ABC, abstractmethod
import math

class shape(ABC):
    @abstractmethod
    def area(self):
        pass

class circle(shape):
    def __init__(self,radius):
        self.radius = radius
    
    def area(self):
        return math.pi * self.radius ** 2

c = circle(4)    
c.area()
```

```python
# Double Inheritance
#Base Class 1
class animal:
    def __init__(self,name,**kwargs):
        self.name = name 
        super().__init__(**kwargs)
        
    def speak(self):
        print("Speaksss")

#Base Class 2
class pet:
    def __init__(self,owner,**kwargs):
        self.owner = owner
        super().__init__(**kwargs)
        
    def owners(self):
       print(f"{self.owner} is the parent of this pet")    
    
#Derived Class
class Dog(animal, pet):
    def __init__(self,name, owner):
        super().__init__(name=name, owner=owner)
    
dog = Dog("lucy", "Awais")
dog.speak()
dog.owners()
```

```python
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img
    def __add__(self, other):
        return Complex(self.real + other.real , self.img + other.img)
    
   
    def __sub__(self, other):
        return Complex(self.real - other.real , self.img - other.img)
    
    def __mul__(self, other):
       real = (self.real * other.real) - (self.img * other.img)
       img = (self.real * other.img) + (self.img * other.real)
       return Complex(real, img)
   
    def __eq__(self, other):
        return self.real == other.real and self.img == other.img
    
    def __repr__(self):
        return f"Complex({self.real} and {self.img}i)"
    
c1 = Complex(7, 10)
c2 = Complex(8, 2)
print(c1 + c2)
```

```python
from abc import ABC,abstractmethod
class Worker(ABC):
    @abstractmethod
    def work(self):
        pass

class Doctor(Worker):
    def work(self):
        print("Hey Im gonna heal you")

class Engineer(Worker):
    def work(self):
        print("Hey Im gonna make the tools")

class Scientist(Doctor, Engineer):
    def work(self):
        Doctor.work(self)
        Engineer.work(self)
    
s = Scientist()
s.work()
```

#### Iterators

```python
class count:
    def __init__(self, num):
        self.num = num
    def __iter__(self):
        return self
    def __next__(self):
       
        if self.num <= 5:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration
                
        

c = count(1)
for x in c:
    print(x)
```

#### Generators

```python
def gen(n):
    for i in range(n):
        yield i**2

x = gen(5)
```

```python
# print(next(x))
for i in x:
    print(i)
```

```python
# Practical Example
def largeFiles(filePath):
    with open(filePath, mode='r') as file:
        for line in file:
            yield line
        
```

```python
fileName = 'data.txt'
for i in largeFiles(fileName):
    print(i.strip())
```

#### Closures

#### Decorator

```python
#Decorator
def mainGreeting(func):
    def subGreeting(*args, **kwargs):
        print("Hello Guys, How are your doing")
        func(*args, **kwargs)
    return subGreeting
```

```python
@mainGreeting
def greet(*args, **kwargs):
    name = args[0]
    age = kwargs['age']
    print(f"Hey {name} What are you upto today, want to learn something new? and your age is {age}")

greet("Awais", age= 22)
```

```python
def openFile( mode):
    def decorator(func):
        def Wrapper(fileName, *args, **kwargs):
            with open(fileName, mode)as f:
                return func(f,*args, **kwargs)
        return Wrapper
    return decorator

@openFile( 'r')
def readtoFile(file):
    content = file.read()
    print(content)
    
readtoFile('data.txt')
```
