```python
###Keywords Arguements

def printDetail(*args, **kwargs):
    for val in args:
        print(f"Positional Arguement: {val}")

    for key,value in kwargs.items():
        print(f"{key} : {value}")


printDetail(1,2,3,4,"Awais",name="Awais", age = 19)
```

```python
### Password Checker
def pass_checker():
    special_character = "!@#$%^&*()"
    spass = input("Enter a strong password")

    if len(spass) >= 8:
        if any(char in special_character for char in spass):
            print("strong password")
        else:
            print("Add atleast one special Character")
    else:
        print("Password must be at least 8 character")

pass_checker()
```

```python
###  Total Cost of Cart

def total_cost(cart):
    total = 0
    for item in cart:
        total += item['Prize'] * item['Quantity']
    return total

cart = [
    {'name': 'Banana', 'Prize': 2, 'Quantity':5},
    {'name': 'Apple', 'Prize': 3, 'Quantity':6}
]

result = total_cost(cart)
print(result)
```

```python
### Palindrome
def is_Palindrome(s):
    s=s.lower().replace(" ", "")
    if s == s[::-1]:
        print("It's a palindrome string")
    else:
        print("not a plaindrome")  

is_Palindrome("Ali")
is_Palindrome("a man a plan a canal panama")
is_Palindrome("Anna")
```

```python
def count_words(file_path):
    word_counts = {}
    with open(file_path, 'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word = word.lower().strip('.,!?;:"\'')
                word_counts[word]= word_counts.get(word,0)+1
    return word_counts

file_path = 'file.txt'
word_frequency = count_words(file_path)
print (word_frequency)
```

```python
def fibo(n):
    a , b = 0 , 1
    for i in range(n):
        if n == 0:
            return 0
        elif n == 1:
            return 1
        else:
            a , b = b , a + b

print(fibo(10))
```

```python
## Counting the Words, Lines & Characters
def count_alphabet_words(file_path):
    with open (file_path, mode='r') as file:
        lines = file.readlines()
        linesCount = len(lines)
        wordCount = sum(len(line.split()) for line in lines)
        charCount = sum(len(line) for line in lines)
    return linesCount, wordCount, charCount


lines, words, character = count_alphabet_words('Awais.txt')
print(f"The Lines are: {lines}, words are {words}, characters are {character}")
```

```python
def read_file(fileName):
    try:
        file = open(fileName, mode='r')
        content = file.read()
    except Exception as e:
        print(e)
    finally:
        try:
            file.closed()
        except Exception as e1:
            pass
```

```python
def sum(lst):
    total = 0
    try:
        for i in lst:
            total += i
        print(total)
    except Exception as e:
         print(f"Error {e}")
    finally:
        print('Program Executed')
        
sum([1,2,3,4,'a']) #Int Type Error
sum([1,2,3,4]) #Execute Smoothly
```

```python
def keyFinder(d, key):
    try:
        value = d[key]
    except KeyError as e:
        print(f"Error {e}")
        value = None
    finally:
        print("Program Executed")
    return value
d = {'name': 'Awais', 'Age': 22}
keyFinder(d, 'Hobbies')
```

```python
def index_checker(lst, index):
    try:
        value = lst[index]
    except Exception as e:
        print(f"Error {e}")
        value = None
    finally:
        print('Program Executed')
    return value

lst = [1,2,3,4,5]
print(index_checker(lst, 10)) 
```

```python
import requests

def readUrl(url):
    try:
        response = requests.get(url)
        response.raise_for_status()
        return response.text
    except requests.RequestException as e:
        print(f'Error {e}')
        return None
    finally:
        print('Program Executed')
print(readUrl('https://jsonplaceholder.typicode.com/posts/1'))
print(readUrl('https://nonexistent.url'))
```
