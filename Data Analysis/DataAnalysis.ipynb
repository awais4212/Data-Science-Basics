#### Numpy Vectorized Opearation

```python
#Vector Opearation
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([1,2,3,4,5])

print(f"Addition: {arr1 + arr2}")
print(f"Substraction: {arr1 - arr2}")
print(f"Multiplication: {arr1 * arr2}")
print(f"Division: {arr1 * arr2}")
```

```python
#Universal Function
arr = np.array([2,4,5,7,8])
print(np.sqrt(arr))
print(np.exp(arr))
print(np.log(arr))
print(np.sin(arr))
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]

# create a line plot
plt.plot(x,y)
plt.xlabel('xPlot')
plt.ylabel('yPlot')
plt.title('Square Graph')
plt.show()
```

```python
## Customized Line Plot
x = [1,2,3,4,5]
y = [1,4,9,16,25]

plt.plot(x,y, color='green',linestyle= '--', marker="o", linewidth='2',markersize='6')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Square Plot')

plt.grid(True)
```

```python
#Multiple plot
x = [1,2,3,4,5]
y1 = [1,4,9,16,25]
y2 = [1,2,3,4,5]
z = [i*2 for i in x ]



plt.figure(figsize=(8,6)) 

plt.subplot(2,2,1)
plt.plot(x,y1,color='green')
plt.title('Plot - 1')


plt.subplot(2,2,2)
plt.plot(x,y2,color='blue', marker='o')


plt.subplot(2,2,3)
plt.plot(z,y1,color='red', marker='o')

plt.subplot(2,2,4)
plt.plot(x,z,color='yellow', marker='o')
```

#### Bar Chart/ Plot


```python
Players = ['Mbappe', 'Messi', 'Ronaldo', 'Haland', 'Kane']
Goals = [8, 8, 3, 7, 6]

# National team jersey colors
colors = [
    '#0055A4',  # Mbappe - France (blue)
    '#75AADB',  # Messi - Argentina (sky blue)
    '#FF0000',  # Ronaldo - Portugal (red)
    '#EF2B2D',  # Haland - Norway (red)
    '#FFFFFF',  # Kane - England (white)
]

plt.bar(Players, Goals, color=colors, edgecolor='black')  # edgecolor so white bar is visible
plt.title("FIFA 2026 TOP SCORER")
plt.xlabel("Players")
plt.ylabel("Goals")
plt.show()
```
