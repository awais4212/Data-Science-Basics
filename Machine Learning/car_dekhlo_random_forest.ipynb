## Car Price Detection

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('data/raw/cardekho_imputated.csv')
```

```python
df.drop(['Unnamed: 0','car_name','brand'],axis=1,inplace=True)
```

```python
df['model'].unique()
```

```python
X=df.drop('selling_price',axis=1)
y=df['selling_price']
```

## Feature Encoding and Standardization

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
X['model']=le.fit_transform(X['model'])
```

```python
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.compose import ColumnTransformer

num_features= X.select_dtypes(exclude='object').columns
one_hot_columns=['seller_type','fuel_type','transmission_type']
label_encoder_column=['model']

numeric_transformer=StandardScaler()
oh_transfromer=OneHotEncoder(drop='first')

preprocessor= ColumnTransformer(
    [
        ("OneHotEncoder",oh_transfromer,one_hot_columns),
        ("StandarScaler",numeric_transformer,num_features)
    ],remainder='passthrough' #only categorical features will change
)
```

```python
X=preprocessor.fit_transform(X)
```

```python
X=pd.DataFrame(X)
```

```python
X
```

## Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

## Model Training

```python
from sklearn.ensemble  import RandomForestRegressor
from sklearn.linear_model import LinearRegression,Lasso,ElasticNet
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error,r2_score,mean_squared_error
```

```python
def model_evaluation(true,prediction):
    mae = mean_absolute_error(true,prediction)
    mse = mean_squared_error(true,prediction)
    rmse = np.sqrt(mse)
    score = r2_score(true,prediction)
    return mae,mse,rmse,score
    
```

```python
models={
     'LinearRegression': LinearRegression(),
     'Lasso': Lasso(),
     'ElasticNet': ElasticNet(),
     'KNN': KNeighborsRegressor(),
     'DecisionTree': DecisionTreeRegressor(),
     'RandomForest':RandomForestRegressor()
}

for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(X_train,y_train)
    
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    
    mae_train,mse_train,rmse_train,rtrain=model_evaluation(y_train,y_train_pred)
    mae_test,mse_test,rmse_test,rtest=model_evaluation(y_test,y_test_pred)
    
    print()
    print(list(models.keys())[i])
    print(f'Mean Square error: {mse_train}')
    print(f'Root Mean Square Error: {rmse_train}')
    print(f'Mean Absolute Error: {mae_train}')
    print(f'R2-Score: {rtrain}')
    print('------------------------')
    
    print('Model Performance on the Test Data')
    print(f'Mean Square error: {mse_test}')
    print(f'Root Mean Square Error: {rmse_test}')
    print(f'Mean Absolute Error: {mae_test}')
    print(f'R2-Score: {rtest}')
    print()
```
