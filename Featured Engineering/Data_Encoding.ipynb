#### Data Encoding

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
```

```python
df = pd.DataFrame({
    "colors": ['red', 'green', 'blue','blue','red', 'green']
})
```

```python
df.head(6)
```

```python
##  Create an instance of encoding 
encoder = OneHotEncoder()
```

```python
encoded=encoder.fit_transform(df[['colors']]).toarray()
```

```python
edf = pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
edf
```

### Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oDF = pd.DataFrame({
    "size":['small','medium','large','small','medium','large']
})
```

```python
ode = OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoded = ode.fit_transform(oDF[['size']])
```

```python
ode.transform([['small']])
```

### Target Guided Ordinal Encoding

```python
import pandas as pd
tdf = pd.DataFrame({
    'city':['khi','lhr','btk', 'khi'],
    'price':[2000,3000,4000,2200]
})
```

```python
tdf
```

```python
from sklearn.preprocessing import TargetEncoder
```

```python
meanPrice=tdf.groupby('city')['price'].mean().to_dict()
```

```python
tdf['encoded_city']=tdf['city'].map(meanPrice)
```
