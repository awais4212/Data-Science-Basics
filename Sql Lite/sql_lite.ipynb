#### SQL Lite 

>SQLite is a lightweight database system that stores all data in a single file and uses SQL commands.


```python
import sqlite3
```

```python
conn = sqlite3.connect('example.db')
conn
```

```python
cursor = conn.cursor()
```

#### Create Query

```python
#### create a table 
cursor.execute('''
Create Table If Not Exists employees(
ID INTEGER Primary Key,
Name TEXT Not Null,
Age INTEGER,
Department Text Not Null
)          
''')

### commit the changes 
conn.commit()
```

```python
#### Insert the Values into the Tables
cursor.execute('''
    Insert into employees(Name,Age, Department)
                values('Awais',22, 'Data Scientist')
''')

cursor.execute('''
    Insert into employees(Name,Age, Department)
                values('Mohid',25, 'Hacker')
''')

cursor.execute('''
    Insert into employees(Name,Age, Department)
                values('lakho',22, 'Networker')
''')

## Commit 
conn.commit()
```

#### Read Query

```python
## Select the particular data from the dataset
cursor.execute('''
    Select * From employees
''')

## use fetchall function 
rows = cursor.fetchall()

## print using the FOR loop 
for row in rows:
    print(row)
```

```python
## reading the data from DB using the Pandas
import pandas as pd
df = pd.read_sql_query("select * from employees", conn)
df
```

#### Update Query

```python
cursor.execute('''
Update employees
Set Age = 21
where Name = "Awais"
''')
```

#### Drop | Delete Query
> Delete used to remove a record
> Whereas, Drop use delete or remove the complete table|

```python
### Drop the Table from the DataBase Query 
cursor.execute('''
DROP TABLE IF EXISTS employees
''')
conn.commit()
```

```python
### Delete The Record From the Database
cursor.execute('''
delete From employees
where name = "Mohid"
''')

conn.commit()
```

#### SQL Project

```python
conn = sqlite3.connect("Sales.db")
cursor = conn.cursor()
```

```python
cursor.execute('''
Create Table if not exists sales(
    Product_ID Integer Primary Key,
    Date Text Not NULL,
    Product Text Not Null,
    Sales Integer,
    Region Text Not Null
)               
''')

sales_data = [
    ('12-2-2023','Bottle', 100, 'Karachi'),
    ('31-7-2025','Spray', 100, 'Karachi'),
    ('11-9-2023','keyboard', 100, 'Karachi'),
    ('1-5-2020','LCD', 100, 'Karachi'),
    ('29-2-2026','Light', 200, 'Islamabad')
]
conn.commit()
```

```python
cursor.executemany('''
insert into sales(Date, Product, Sales, Region)
    values(?,?,?,?)
''',sales_data)
conn.commit()
```

```python
cursor.execute('''
    select * From sales
''')

rows = cursor.fetchall()
for row in rows:
    print(row)
conn.commit()
```
