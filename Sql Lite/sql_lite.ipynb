#### SQL Lite 

>SQLite is a lightweight database system that stores all data in a single file and uses SQL commands.


```python
import sqlite3
```

```python
conn = sqlite3.connect('example.db')
conn
```

```python
cursor = conn.cursor()
```

#### Create Query

```python
#### create a table 
cursor.execute('''
Create Table If Not Exists employees(
ID INTEGER Primary Key,
Name TEXT Not Null,
Age INTEGER,
Department Text Not Null
)          
''')

### commit the changes 
conn.commit()
```

```python
#### Insert the Values into the Tables
cursor.execute('''
    Insert into employees(Name,Age, Department)
                values('Awais',22, 'Data Scientist')
''')

cursor.execute('''
    Insert into employees(Name,Age, Department)
                values('Mohid',25, 'Hacker')
''')

cursor.execute('''
    Insert into employees(Name,Age, Department)
                values('lakho',22, 'Networker')
''')

## Commit 
conn.commit()
```

#### Read Query

```python
## Select the particular data from the dataset
cursor.execute('''
    Select * From employees
''')

## use fetchall function 
rows = cursor.fetchall()

## print using the FOR loop 
for row in rows:
    print(row)
```

```python
## reading the data from DB using the Pandas
import pandas as pd
df = pd.read_sql_query("select * from employees", conn)
df
```

#### Update Query

```python
cursor.execute('''
Update employees
Set Age = 21
where Name = "Awais"
''')
```

#### Drop | Delete Query
> Delete used to remove a record
> Whereas, Drop use delete or remove the complete table|

```python
### Drop the Table from the DataBase Query 
cursor.execute('''
DROP TABLE IF EXISTS employees
''')
conn.commit()
```

```python
### Delete The Record From the Database
cursor.execute('''
delete From employees
where name = "Mohid"
''')

conn.commit()
```

### Assignment 6: Advanced Queries

1. Write a Python function to fetch and display employees older than a certain age.

```python
def fetchbyAge(age):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('''
    select * From employees
    where Age > ?
    ''',(age,))
    rows = cursor.fetchall()
    for row in rows:
        print(row)
    conn.commit()
    conn.close()
```
