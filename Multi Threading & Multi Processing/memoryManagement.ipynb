#### Python Memory Management
```Python handles memory automatically, but knowing the mechanics helps you debug leaks and write efficient code.```

````python
import sys

a=[]
print(sys.getrefcount(a))
````

````python
b = a
sys.getrefcount(b)
````

````python
try:
    del b
    sys.getrefcount(b)
except Exception as e:
    print(f"The error occured due to the {e}")
````

#### Garbage collector (for cycles)
```Reference counting alone can't handle circular references (e.g., a.ref = b; b.ref = a). Python's gc module periodically scans for and cleans up these cycles.```

````python
import gc

#enable
gc.enable()
````

````python
gc.disable()
````

````python
gc.collect()
````

````python
gc.get_stats()
````

````python
print(gc.garbage)
````

#### Memory Management Best practise

````python
#Handled Circular reference
import gc

class myobject:
    def __init__(self,name):
        self.name = name
        print(f"Object is Created {self.name}")
        
    def __del__(self):
        print(f"Object {self.name} is deleted")
    
obj1 = myobject('obj1')
obj2 = myobject('obj2')

obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2

# Manually Trigger the Garbage Collection
gc.collect()
````

````python
def generateNumbers(n):
    for i in range(n):
        yield i
    
for num in generateNumbers(100000):
    print(num)
    if num>10:
        break
````
