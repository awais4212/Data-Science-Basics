## Car Prediction Neural Network

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf # Handling with Deep Learning Models
````

````python
tensor_zero_d=tf.constant(0)
print(tensor_zero_d)
print(tensor_zero_d.ndim)
````

````python
tensor_one_d=tf.constant([0,1,2])
print(tensor_one_d)
print(tensor_one_d.ndim)
````

````python
tensor_two_d=tf.constant([[0,1,2],
                           [4,5,6]])
print(tensor_two_d)
print(tensor_two_d.ndim)
````

````python
tensor_three_d=tf.constant([[[0,1,2],
                           [4,5,6]],
                          [[0,1,2],
                           [4,5,6]],])
print(tensor_three_d)
print(tensor_three_d.ndim)
````

## Tensor Random_Normal
```This Distributes Randomely by using Normal Distribution```

````python
random_normal_tensor=tf.random.normal([2,3],stddev=1.0,mean=0.0,seed=42)
print(random_normal_tensor)
````

## Tensor Random_Uniform
```This Distributes Randomely by using Uniform Distribution```

````python
random_uniform_tensor=tf.random.uniform(
    [3,3],
    minval=0,
    maxval=100,
    dtype=tf.dtypes.float32,
    seed=None,
    name=None
)
print(random_uniform_tensor)
````

## Random Seed
```Generated Random Numbers will not changes``` 

````python
tf.random.set_seed(5)
print(tf.random.uniform(shape=[3,],maxval=5, dtype=tf.int32,seed=10))
print(tf.random.uniform(shape=[3,],maxval=5, dtype=tf.int32,seed=10))
print(tf.random.uniform(shape=[3,],maxval=5, dtype=tf.int32,seed=10))
````

## Reduce Tensor
```Will Reduce the metrics into one Value```

````python
x=tf.constant([1,2,3,4,5])
reduce_tensor=tf.math.reduce_max(x)
print(reduce_tensor)
````

````python
# Sum of the Tensor
x=tf.constant([1,2,3,4,5])
reduce_tensor=tf.math.reduce_sum(x,keepdims=False)
reduce_tensor1=tf.math.reduce_sum(x,keepdims=True)
print(reduce_tensor)
print(reduce_tensor1)
````

````python
# sigmoid function
x=tf.constant([0.0,-1.2,0.9,1.5,50.0])
tensor_sigmoid=tf.math.sigmoid(x)
print(tensor_sigmoid)
````
