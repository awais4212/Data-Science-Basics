## Car Prediction Neural Network

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf # Handling with Deep Learning Models
````

````python
tensor_zero_d=tf.constant(0)
print(tensor_zero_d)
print(tensor_zero_d.ndim)
````

````python
tensor_one_d=tf.constant([0,1,2])
print(tensor_one_d)
print(tensor_one_d.ndim)
````

````python
tensor_two_d=tf.constant([[0,1,2],
                           [4,5,6]])
print(tensor_two_d)
print(tensor_two_d.ndim)
````

````python
tensor_three_d=tf.constant([[[0,1,2],
                           [4,5,6]],
                          [[0,1,2],
                           [4,5,6]],])
print(tensor_three_d)
print(tensor_three_d.ndim)
````

## Converting String to Tensors

````python
tensor_string=tf.constant(['hello','hi'])
print(tensor_string)
````

## Converting Boolean to Tensors

````python
tensor_bool=tf.constant([True,True,False])
print(tensor_bool)
````

## Converting Numpy Array Into Tensor 

````python
arr=np.array([1,2,3,4,5])
tensor_array=tf.convert_to_tensor(arr)
print(tensor_array)
````

## Eye Tensor
```Parameters: num_columns, num_rows, batch_shape```


````python
eye=tf.eye(num_columns=None,num_rows=3,batch_shape=[4,])
print(eye)
````

## Fill Tensor
```Parameters: dim, value, name```

````python
fill_tensor=tf.fill(
    [3,3],5
)
print(fill_tensor)
````

## One Tensor
```Parameters: shape```

````python
one_tensors=tf.ones([1,3,3])
print(one_tensors)
````

## Ones_Like Tensor

````python
ones_like_tensor=tf.ones_like(fill_tensor)
print(ones_like_tensor)
````

## Zero Tensors

````python
zero_tensors=tf.zeros([1,3,3])
print(zero_tensors)
````

## Zeros_Like Tensors

````python
zero_tensors=tf.zeros_like(fill_tensor)
print(zero_tensors)
````

## Tensor Shape

````python
tensor_shape=tf.shape(tensor_three_d)
print(tensor_shape)
````

## Tensor Rank

````python
tensor_one_rank= tf.rank(tensor_one_d)
tensor_two_rank= tf.rank(tensor_two_d)
tensor_three_rank= tf.rank(tensor_three_d)
print(tensor_one_rank)
print(tensor_two_rank)
print(tensor_three_rank)
````

## Tensor Size
```Return Total Numbers of Elements in the Tensor```

````python
tensor_size=tf.size(tensor_three_d)
print(tensor_size)
````

## Tensor Random_Normal
```This Distributes Randomely by using Normal Distribution```

````python
random_normal_tensor=tf.random.normal([2,3],stddev=1.0,mean=0.0,seed=42)
print(random_normal_tensor)
````

## Tensor Random_Uniform
```This Distributes Randomely by using Uniform Distribution```

````python
random_uniform_tensor=tf.random.uniform(
    [3,3],
    minval=0,
    maxval=100,
    dtype=tf.dtypes.float32,
    seed=None,
    name=None
)
print(random_uniform_tensor)
````

## Random Seed
```Generated Random Numbers will not changes``` 

````python
tf.random.set_seed(5)
print(tf.random.uniform(shape=[3,],maxval=5, dtype=tf.int32,seed=10))
print(tf.random.uniform(shape=[3,],maxval=5, dtype=tf.int32,seed=10))
print(tf.random.uniform(shape=[3,],maxval=5, dtype=tf.int32,seed=10))
````
