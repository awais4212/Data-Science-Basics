## Linear ALgebra

## Matrix Multiplication
```Rows & Columns should be equal to each other of Matrix A and Matrix B```

````python
a=tf.constant([[1,3,2,6,7],
               [1,2,1,3,2],
               [4,3,1,3,1]])
b=tf.constant([[1,3,2,6,7],
               [1,2,1,3,2],
               [1,2,1,3,2],
               [1,2,1,3,2],
               [4,3,1,3,1]] )
print(a.shape,b.shape)
tf.linalg.matmul(a,b)
````

## For Rows and Column Wise multiplication use ``` @ ```
``` Input @ Weights + Bias ---> Output ```

````python
a@b
````

## Transpose Of Matrix

````python
a=tf.constant([[1,2,1],
                [2,3,1]])
tf.transpose(a)
````

````python
tensor=tf.constant([[1,2,3],
                    [2,3,1],
                    [2,1,2]])
print(tf.linalg.band_part(tensor, 0, -1))  # upper triangular
print(tf.linalg.band_part(tensor, -1, 0))  # lower triangular
print(tf.linalg.band_part(tensor, 0, 0))
````

## Common Methods

#### Expand Dimension ```expand_dim```
```It will Increase the Dimsion by 1```

````python
a=tf.constant([1,2,3,4,5])
print(a.shape)
````

````python
# Row-Wise
tensor_expand_dim=tf.expand_dims(a,axis=0)
print(tensor_expand_dim)
````

````python
# Column-Wise
tensor_expand_dim=tf.expand_dims(a,axis=1)
print(tensor_expand_dim)
````

## Tensor Squeeze
```Back The Expand Tensor to the actual dimension```

````python
## Squeeze
orignal_tensor=tf.squeeze(tensor_expand_dim)
print(orignal_tensor)
````

#### Reshape() ```tf.reshape(tensor,shape)```
```Give The New Shapes to the Tensors```

````python
a=tf.constant([1,2,3,4])
tf.reshape(a,shape=[2,2])
````

````python
### If don't have right idea how to many the values are present in the tensors use -1
a=tf.constant([1,2,3,4,1,2,2,2])
tf.reshape(a,shape=[2,-1])
````

#### Concat ```tf.concat([tensor1,tensor2],axis)```

````python
a=tf.constant([[1,2,3,4],
               [4,5,6,7]])
b=tf.constant([[0,0,0,0],
               [1,1,1,1]])
concat=tf.concat([a,b],axis=0)
print(concat)
````

````python
a=tf.constant([[1,2,3,4],
               [4,5,6,7]])
b=tf.constant([[0,0,0,0],
               [1,1,1,1]])
concat=tf.concat([a,b],axis=1)
print(concat)
````

### Stack

````python
s=tf.stack([a,b],axis=0)
print(s)
````

#### Padding ```pad(tensor,padding,constant)```
```Padding use for the image processing in CNN```

````python
image=tf.constant([[1,1,1],
                   [1,1,1]])
padding=tf.constant([[1,1],
                     [2,4]])
print(tf.pad(image,padding,"CONSTANT"))
````

````python
image=tf.constant([
                   [[1,1,1],
                   [1,1,1],
                   [1,1,1]]
                   ])
padding=tf.constant([[0,0],
                     [1,1],
                     [1,1]])

print(tf.pad(image,padding,"CONSTANT",constant_values=0))
````
